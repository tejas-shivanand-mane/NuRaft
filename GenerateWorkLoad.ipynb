```python
import random

def generate_workload(write_percentage, num_txns, output_file="commands.txt"):
    write_count = int((write_percentage / 100.0) * num_txns)
    read_count = num_txns - write_count

    # Divide writes evenly (or almost evenly) between push (0) and pop (1)
    push_count = write_count // 2
    pop_count = write_count - push_count

    commands = (["msg 0"] * push_count) + (["msg 1"] * pop_count) + (["msg 2"] * read_count)
    random.shuffle(commands)

    with open(output_file, "w") as f:
        for cmd in commands:
            f.write(cmd + "\n")

    print(f"Workload written to {output_file} — {push_count} push, {pop_count} pop, {read_count} read")

```

```python
write_pct, num_txns, output_file = 15,  500000, 'commands.txt'

generate_workload(write_pct, num_txns, output_file)
```
